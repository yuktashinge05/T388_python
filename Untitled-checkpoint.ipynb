# WAP to print whether a number is an armstrong number

```python
num=int(input("enter any number to check whether a armstrong or not: "))

length= len(str(num))
result= 0
for i in str (num):
    result= result + (int(i)**length)
if result == num:
    print("the number is armstrong number ! ")
else:
    print("oh!, this number is not an armstrong number.")
```

# factorial 

```python
# 5!=5*4*3*2*1
def fact(n):
    if n==1:
        return 1
    else:
        product = n * fact(n-1)
        return(product)
fact(5)
```
